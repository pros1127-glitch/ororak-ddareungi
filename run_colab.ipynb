## (선택) 원본 대여이력부터 직접 집계하기
서울 열린데이터광장에서 받은 `서울특별시 공공자전거 대여이력 정보_2606.csv`(또는 zip)를 구글 드라이브 `공공데이터` 폴더에 올린 뒤 실행합니다(5~10분).

```python
from google.colab import drive
drive.mount('/content/drive')
import glob, os, zipfile, shutil
DATA_DIR = '/content/drive/MyDrive/공공데이터'          # ← 드라이브 속 폴더 경로
os.makedirs('data/trips', exist_ok=True)
zips = glob.glob(f'{DATA_DIR}/*대여이력*.zip'); csvs = glob.glob(f'{DATA_DIR}/*대여이력*.csv')
if zips:
    zipfile.ZipFile(zips[0]).extractall('data/trips')
elif csvs:
    shutil.copy(csvs[0], 'data/trips/')
print(glob.glob('data/trips/*.csv'))
```
